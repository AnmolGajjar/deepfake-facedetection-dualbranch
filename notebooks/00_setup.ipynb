```python
import torch
import timm
import cv2
import albumentations
import sklearn
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ── System info ──────────────────────────────
print("=" * 45)
print("  Environment Check")
print("=" * 45)
print(f"  Python     : 3.13")
print(f"  PyTorch    : {torch.__version__}")
print(f"  CUDA       : {torch.version.cuda}")
print(f"  NumPy      : {np.__version__}")
print(f"  timm       : {timm.__version__}")
print(f"  OpenCV     : {cv2.__version__}")

# ── GPU ──────────────────────────────────────
print("\n" + "=" * 45)
print("  GPU Check")
print("=" * 45)
assert torch.cuda.is_available(), "GPU NOT FOUND"
props = torch.cuda.get_device_properties(0)
print(f"  GPU        : {props.name}")
print(f"  VRAM       : {props.total_memory/1024**3:.1f} GB")

# ── Tensor on GPU ────────────────────────────
x = torch.randn(4, 3, 224, 224).cuda()
print(f"  Tensor     : {x.shape} on {x.device}")

# ── EfficientNet-B0 forward pass ─────────────
print("\n" + "=" * 45)
print("  Model Check")
print("=" * 45)
model = timm.create_model('efficientnet_b0', pretrained=True).cuda().eval()
with torch.no_grad():
    out = model(x)
print(f"  EfficientNet-B0 output : {out.shape}")

# ── FFT check ────────────────────────────────
gray = x[0].mean(dim=0)
fft  = torch.fft.fftshift(torch.fft.fft2(gray))
mag  = torch.log1p(torch.abs(fft))
print(f"  FFT spectrum shape     : {mag.shape}")

print("\n" + "=" * 45)
print("  ALL CHECKS PASSED")
print("  Ready for Step 2 — Dataset Loading")
print("=" * 45)
```
